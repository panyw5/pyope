# rank-1 $\mathcal{N} = 3$ theory (in Gabi's paper) associated to $\mathbb{Z}_3$ group

```python
# 定义自由场算符
# bc 系统（鬼场）：b 权重 2, c 权重 -1
clear_registry()
b = BasisOperator('b', fermionic=True, conformal_weight=Fraction(2))
c = BasisOperator('c', fermionic=True, conformal_weight=Fraction(-1))

# βγ 系统：β 权重 3/2, γ 权重 -1/2
β = BasisOperator('β',  conformal_weight=Fraction(3, 2))
γ = BasisOperator('γ',  conformal_weight=Fraction(-1, 2))
```

```python
OPE[b, c] = MakeOPE([One])
OPE[β, γ] = MakeOPE([-One])
```

```python
J = 2 * NO(b, c) + 3 * NO(β, γ)

G = NO(b, γ)
Gbar = 2 * NO(d(β), c) + 3 * NO(β, d(c))

T = - 2 * NO(b, d(c)) - Fraction(3,2) * NO(β, d(γ)) - NO(d(b), c) - Fraction(1,2) * NO(d(β), γ)


W = β  # 权重 3/2
GW = b    # 权重 2

Wbar = (NO(β, NO(β, NO(γ, NO(γ, γ))))
      + 2 * NO(β, NO(γ, NO(γ, NO(b, c))))
      - 4 * NO(β, NO(d(γ), γ))
      - Fraction(4, 3) * NO(γ, NO(b, d(c)))
      + Fraction(2, 3) * NO(γ, NO(d(b), c))
      + Fraction(2, 3) * NO(d(β), NO(γ, γ))
      - Fraction(8, 3) * NO(d(γ), NO(b, c))
      + Fraction(10, 9) * d(d(γ)))

GbarWbar = (Fraction(8, 3) * NO(b, NO(d(d(c)), c))
            + 3 * NO(β, NO(β, NO(γ, NO(γ, d(c)))))
            - 4 * NO(β, NO(γ, NO(b, NO(d(c), c))))
            - 4 * NO(β, NO(γ, d(d(c))))
            - 4 * NO(β, NO(d(γ), d(c)))
            - Fraction(2, 3) * NO(d(b), NO(d(c), c))
            + 2 * NO(d(β), NO(β, NO(γ, NO(γ, c))))
            - Fraction(8, 3) * NO(d(β), NO(d(γ), c))
            + Fraction(2, 3) * NO(d(d(β)), NO(γ, c))
            + Fraction(10, 9) * d(d(d(c))))

generators = [T, W, J, G, Gbar, GW, Wbar, GbarWbar]
```

```python
result1 = NO(T, NO(J, J))
simplify_with_wolfram(result1)
```

```python
simplify_with_wolfram(- Fraction(8, 9) * NO(T, NO(T, NO(T, NO(J, J))))
+ Fraction(4, 3) * NO(T, NO(T, NO(T, T)))
+ Fraction(4, 9) * NO(T, NO(T, NO(T, d(J))))
+ Fraction(-8, 9) * NO(T, NO(T, NO(J, NO(G, Gbar))))
+ Fraction(2, 3) * NO(T, NO(T, NO(G, d(Gbar))))
+ Fraction(-3) * NO(T, NO(T, NO(W, d(Wbar))))
+ Fraction(-1, 9) * NO(T, NO(T, NO(d(J), NO(J, J))))
+ Fraction(-4, 9) * NO(T, NO(T, NO(d(J), d(J))))
+ Fraction(-2, 9) * NO(T, NO(T, NO(d(G), Gbar)))
+ Fraction(5) * NO(T, NO(T, NO(d(W), Wbar)))
+ Fraction(1) * NO(T, NO(T, NO(d(J, 2), J))))
```

```python
def flatten_and_deduplicate(lst):
    def flatten_completely(lst):
        result = []
        for item in lst:
            if isinstance(item, list):
                result.extend(flatten_completely(item))
            else:
                result.append(item)
        return result
    flatten = flatten_completely(lst)

    unique = []
    for item in flatten:
        if item not in unique:
            unique.append(item)
    return unique
```

```python
jacobi_identities = [[check_jacobi_identity(GWbar, g2, g3)
                     for g2 in generators] for g3 in generators]


non_zero = flatten_and_deduplicate(jacobi_identities)
[simplify(expr) for expr in non_zero]
```

# $U(1)$ $\mathcal{N} = 4$ theory + $\mathbb{Z}_3$ gauging

```python
b = BasisOperator('b', conformal_weight=1, fermionic=True)
c = BasisOperator('c', conformal_weight=0, fermionic=True)
β = BasisOperator('β', conformal_weight=1/2)
γ = BasisOperator('γ', conformal_weight=1/2)
```

```python
OPE[β, γ] = MakeOPE([-One])
OPE[b, c] = MakeOPE([One])
T = (1/2) * (NO(d(β), γ) - NO(β, d(γ))) - NO(b, d(c))
OPE(T, b).pole(2) - b
OPE(T, c).pole(2)
OPE(T, β).pole(2) - (1/2) * β
OPE(T, γ).pole(2) - (1/2) * γ

# check Virasoro algebra
OPE(T,T).pole(4)
OPE(T, T).pole(3)
OPE(T, T).pole(2) - 2 * T
simplify(OPE(T, T).pole(1) - d(T))
```
