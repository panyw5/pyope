# rank-1 $\mathcal{N} = 3$ theory (in Gabi's paper) associated to $\mathbb{Z}_3$ group

```python
# 定义自由场算符
# bc 系统（鬼场）：b 权重 2, c 权重 -1
b = BasisOperator('b', fermionic=True, conformal_weight=Fraction(2))
c = BasisOperator('c', fermionic=True, conformal_weight=Fraction(-1))

# βγ 系统：β 权重 3/2, γ 权重 -1/2
β = BasisOperator('β',  conformal_weight=Fraction(3, 2))
γ = BasisOperator('γ',  conformal_weight=Fraction(-1, 2))
OPE[b, c] = MakeOPE([One])
OPE[β, γ] = MakeOPE([-One])
```

```python
J = 2 * NO(b, c) + 3 * NO(β, γ)

G = NO(γ, b)
Gt = 2 * NO(d(β), c) + 3 * NO(β, d(c))

T = - 2 * NO(b, d(c)) - (3/2) * NO(β, d(γ)) - NO(d(b), c) - (1/2) * NO(d(β), γ)


W = β  # 权重 3/2
GW = b    # 权重 2



Wbar = (NO(β, NO(β, NO(γ, NO(γ, γ))))
      + 2 * NO(β, NO(γ, NO(γ, NO(b, c))))
      - 4 * NO(β, NO(d(γ), γ))
      - Fraction(4, 3) * NO(γ, NO(b, d(c)))
      + Fraction(2, 3) * NO(γ, NO(d(b), c))
      + Fraction(2, 3) * NO(d(β), NO(γ, γ))
      - Fraction(8, 3) * NO(d(γ), NO(b, c))
      + Fraction(10, 9) * d(d(γ)))

GWbar = (Fraction(8, 3) * NO(b, NO(d(d(c)), c))
       + 3 * NO(β, NO(β, NO(γ, NO(γ, d(c)))))
       - 4 * NO(β, NO(γ, NO(b, NO(d(c), c))))
       - 4 * NO(β, NO(γ, d(d(c))))
       - 4 * NO(β, NO(d(γ), d(c)))
       - Fraction(2, 3) * NO(d(b), NO(d(c), c))
       + 2 * NO(d(β), NO(β, NO(γ, NO(γ, c))))
       - Fraction(8, 3) * NO(d(β), NO(d(γ), c))
       + Fraction(2, 3) * NO(d(d(β)), NO(γ, c))
       + Fraction(10, 9) * d(d(d(c))))

generators = [T, W, J, G, Gt, GW, Wbar, GWbar]

(T, W, J, G, Gt, GW, Wbar, GWbar) = make_realized(generators)
generators = [T, W, J, G, Gt, GW, Wbar, GWbar]
```

```python
def flatten_and_deduplicate(lst):
    def flatten_completely(lst):
        result = []
        for item in lst:
            if isinstance(item, list):
                result.extend(flatten_completely(item))
            else:
                result.append(item)
        return result
    flatten = flatten_completely(lst)

    unique = []
    for item in flatten:
        if item not in unique:
            unique.append(item)
    return unique
```

```python
jacobi_identities = [[check_jacobi_identity(GWbar, g2, g3)
                     for g2 in generators] for g3 in generators]


non_zero = flatten_and_deduplicate(jacobi_identities)
[simplify(expr) for expr in non_zero]
```

```python
# 用 generators 所生成的算符空间
weight=3
basis = LocalOperatorBasis(generators, max_weight=weight)


# 用 b, c, β, γ 所生成的算符空间
bc_basis = LocalOperatorBasis([b, c, β, γ], max_weight=weight, max_occurence=6)

# 列举 weight=4 的算符，有 10 个
list_of_ops = [op for op in basis.list(weight=weight)]
list_of_ops_realized = [op.realize() for op in basis.list(weight=weight)]
len(list_of_ops)

bc_basis.list_independent_ops(list_of_ops_realized)
relations = bc_basis.list_zero_relations(list_of_ops_realized)
```

```python
null = sum([relations[0]["coefficients"][i] * list_of_ops[i] for i in range(len(list_of_ops_realized))])
```

```python
weights = [0, 1, 2, Fraction(5, 2), 3, Fraction(7, 2), 4, Fraction(9, 2), 5]
free_field_basis = LocalOperatorBasis(
    [b, c, β, γ], max_weight=max(weights), max_occurence=10)
generator_basis = LocalOperatorBasis(generators, max_weight=max(weights))

basis_count_data = []
basis_examples = {}

for weight in weights:
    basis_at_weight = generator_basis.list(weight)
    bosonic_basis = [
        op for op in basis_at_weight if get_operator_parity(op) == 0]
    fermionic_basis = [
        op for op in basis_at_weight if get_operator_parity(op) == 1]

    bosonic_realized = [
        expr for expr in (realize_and_simplify(op) for op in bosonic_basis) if expr != 0
    ]
    fermionic_realized = [
        expr for expr in (realize_and_simplify(op) for op in fermionic_basis) if expr != 0
    ]

    independent_bosonic = free_field_basis.list_independent_ops(
        bosonic_realized, weight=weight, max_occurence=10)
    independent_fermionic = free_field_basis.list_independent_ops(
        fermionic_realized, weight=weight, max_occurence=10)

    basis_count_data.append({
        'weight': weight,
        'bosonic_count': len(independent_bosonic),
        'fermionic_count': len(independent_fermionic),
        'supertrace': len(independent_bosonic) - len(independent_fermionic),
    })

    basis_examples[weight] = {
        'bosonic': bosonic_basis,
        'fermionic': fermionic_basis,
    }

basis_count_data
```

```python
for row in basis_count_data:
    print(
        f"h = {row['weight']}: bosonic = {row['bosonic_count']}, fermionic = {row['fermionic_count']}, supertrace = {row['supertrace']}"
    )

supercharacter = ' + '.join(
    [f"{row['supertrace']} q^{row['weight']}" for row in basis_count_data]
)
supercharacter
```

```python
OPE[β, γ] = MakeOPE([-One])
OPE[b, c] = MakeOPE([One])
```

```python
J = NO(β, γ)
T = (1/2) * (NO(d(β), γ) - NO(β, d(γ))) - NO(b, d(c))
G = NO(b, γ)
Gbar = NO(β, d(c))
W = NO(β, NO(β, β))
Wbar = NO(γ, NO(γ, γ))
GWplus = NO(β, NO(β, b))
GWminus = NO(γ, NO(γ, d(c)))

generators = [T, J, G, Gbar, W, Wbar, GWplus, GWminus]

T, J, G, Gbar, W, Wbar, GWplus, GWminus = make_realized(generators)
generators = [T, J, G, Gbar, W, Wbar, GWplus, GWminus]
```

```python
def flatten_and_deduplicate(lst):
    def flatten_completely(lst):
        result = []
        for item in lst:
            if isinstance(item, list):
                result.extend(flatten_completely(item))
            else:
                result.append(item)
        return result
    flatten = flatten_completely(lst)

    unique = []
    for item in flatten:
        if item not in unique:
            unique.append(item)
    return unique
```

## $\epsilon \in 3\mathbb{Z}$ sector generated by $b, \partial c, \beta, \gamma$

下面统计的态空间不是由 $b,c,\beta,\gamma$ 生成，而是由 $b,\partial c,\beta,\gamma$ 生成。

因此裸的 $c$ 不允许出现，例如 $\beta c$ 不在这个态空间里。这里使用的规则是 $\epsilon(\partial A)=\epsilon(A)$，以及 $\epsilon(\mathrm{NO}(A,B))=\epsilon(A)+\epsilon(B)$。


```python
from pyope.operators import Operator, DerivativeOperator, NormalOrderedOperator
from pyope.local_operator import extract_scalar_operator

def epsilon_of(expr):
    if expr == One or expr == Zero:
        return 0
    if isinstance(expr, DerivativeOperator):
        return epsilon_of(expr.base)
    if isinstance(expr, NormalOrderedOperator):
        return epsilon_of(expr.left) + epsilon_of(expr.right)
    if isinstance(expr, Operator):
        return getattr(expr, 'epsilon', 0)
    if isinstance(expr, sp.Add):
        epsilons = {epsilon_of(arg) for arg in expr.args}
        if len(epsilons) != 1:
            raise ValueError(f'inconsistent epsilon in sum: {epsilons}')
        return next(iter(epsilons)) if epsilons else 0
    if isinstance(expr, sp.Mul):
        _, op = extract_scalar_operator(expr)
        return epsilon_of(op)
    if isinstance(expr, sp.Expr) and not expr.has(Operator):
        return 0
    raise TypeError(f'Cannot determine epsilon of {expr!r}')

epsilon_weights = [0, Fraction(1, 2), 1, Fraction(3, 2), 2, Fraction(5, 2), 3, Fraction(7, 2), 4, Fraction(9, 2), 5, Fraction(11, 2), 6]
epsilon_basis = LocalOperatorBasis([b, d(c), β, γ], max_weight=max(epsilon_weights), max_occurence=10)
epsilon_mod_3_data = []
epsilon_mod_3_examples = {}

for weight in epsilon_weights:
    basis_at_weight = epsilon_basis.list(weight, max_occurence=10)
    epsilon_zero_mod_3 = [op for op in basis_at_weight if epsilon_of(op) % 3 == 0]
    bosonic_basis = [op for op in epsilon_zero_mod_3 if get_operator_parity(op) == 0]
    fermionic_basis = [op for op in epsilon_zero_mod_3 if get_operator_parity(op) == 1]

    epsilon_mod_3_data.append({
        'weight': weight,
        'bosonic_count': len(bosonic_basis),
        'fermionic_count': len(fermionic_basis),
        'supertrace': len(bosonic_basis) - len(fermionic_basis),
    })

    epsilon_mod_3_examples[weight] = {
        'bosonic': bosonic_basis,
        'fermionic': fermionic_basis,
    }

epsilon_mod_3_data
```

```python
for row in epsilon_mod_3_data:
    print(
        f"h = {row['weight']}: bosonic = {row['bosonic_count']}, fermionic = {row['fermionic_count']}, supertrace = {row['supertrace']}"
    )

epsilon_mod_3_supercharacter = ' + '.join(
    [f"{row['supertrace']} q^{row['weight']}" for row in epsilon_mod_3_data]
)
epsilon_mod_3_supercharacter
```

```python
for weight in [0, Fraction(1, 2), 1, Fraction(3, 2), 2]:
    print(f'h = {weight}')
    print('  bosonic:')
    for op in epsilon_mod_3_examples[weight]['bosonic']:
        print(f'    {op}')
    print('  fermionic:')
    for op in epsilon_mod_3_examples[weight]['fermionic']:
        print(f'    {op}')
    print()
```

因此在 $\epsilon \equiv 0 \pmod 3$ 扇区里，前几项超特征标为

$$
\operatorname{tr}_{\epsilon \equiv 0 \; (3)} (-1)^F q^{L_0-c/24} = q^{-c/24}(1 + q + 2 q^2 - 2 q^{5/2} + 3 q^3 - 2 q^{7/2} + 4 q^4 - 4 q^{9/2} + 6 q^5 - 6 q^{11/2} + 9 q^6 + \cdots).
