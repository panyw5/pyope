```python
# 导入必要的库
import itertools
from fractions import Fraction
from pyope import BasisOperator, OPE, simplify, expand_nested_no, NO, d, Bosonic, Fermionic, One, Zero
from sympy import Rational, Symbol
import sys
sys.path.insert(0, '../src')
```

```python
J_plus = BasisOperator("J⁺", conformal_weight=1)
J_zero = BasisOperator("J⁰", conformal_weight=1)
J_minus = BasisOperator("J⁻", conformal_weight=1)

k_val = 1  # 使用具体值

OPE[J_plus, J_zero] = OPE.make([-2 * J_plus])
OPE[J_plus, J_minus] = OPE.make([k_val * One, J_zero])
OPE[J_zero, J_minus] = OPE.make([-2 * J_minus])
OPE[J_zero, J_zero] = OPE.make([2 * k_val * One, 0])

```

```python
J = BasisOperator("J", conformal_weight=1)
Bosonic(J)
k = Symbol("k", positive=True)
OPE[J, J] = OPE.make([k * One, 0])
T = NO(J, J) / (2 * k)
result = OPE(T, T)  # 计算 OPE(T, T)
# 问题：result.pole(2) 返回 J**2/k，而不是 NO(J,J)/k
pole_2 = result.pole(2)
print('pole_2 =', pole_2)  # 输出: J**2/k
```
