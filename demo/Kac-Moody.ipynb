```python
J_plus = BasisOperator("J⁺", conformal_weight=1)
J_zero = BasisOperator("J⁰", conformal_weight=1)
J_minus = BasisOperator("J⁻", conformal_weight=1)
Bosonic(J_plus, J_zero, J_minus)
```

```python
k_val = 1

OPE[J_plus, J_zero] = OPE.make([-2 * J_plus])
OPE[J_plus, J_minus] = OPE.make([k_val * One, J_zero])
OPE[J_zero, J_minus] = OPE.make([-2 * J_minus])
OPE[J_zero, J_zero] = OPE.make([2 * k_val * One, 0])

T = (NO(J_plus, J_minus) + NO(J_minus, J_plus) + NO(J_zero, J_zero) / 2) / (
    2 * (k_val + 2)
)
```

```python
[
  simplify(OPE(T, T).pole(1)-d(T)),
  simplify(OPE(T, T).pole(2) - 2 * T),
  simplify(OPE(T, T).pole(3)),
  simplify(OPE(T, T).pole(4) - Rational(1, 2) * One),
]
```
