# normal_product() 函数演示

本 notebook 演示如何使用 `normal_product()` 函数将多个算符的乘积转化为嵌套的正规序。

## 功能说明

`normal_product(A, B, C, ...)` 将算符乘积 $A \cdot B \cdot C \cdots$ 转化为嵌套的正规序：

$$\text{normal_product}(A, B, C, D) = (A(B(CD)))$$

即 `NO(A, NO(B, NO(C, D)))`

```python
import sympy as sp
from pyope import (
    BasicOperator,
    NO,
    normal_product,
    d,
    One,
    Zero,
)

sp.init_printing()
```

## 1. 基本用法

```python
# 定义一些基本算符
T = BasicOperator("T", 2)
J = BasicOperator("J", 1)
W = BasicOperator("W", 3)

print(f"T: {T}")
print(f"J: {J}")
print(f"W: {W}")
```

### 两个算符

```python
result = normal_product(T, J)
print(f"normal_product(T, J) = {result}")
print(f"等价于: NO(T, J) = {NO(T, J)}")
print(f"相等: {result == NO(T, J)}")
```

### 三个算符

```python
result = normal_product(T, J, W)
print(f"normal_product(T, J, W) = {result}")
print(f"等价于: NO(T, NO(J, W)) = {NO(T, NO(J, W))}")
print(f"相等: {result == NO(T, NO(J, W))}")
```

### 四个算符

```python
L = BasicOperator("L", 1)

result = normal_product(T, J, W, L)
print(f"normal_product(T, J, W, L) = {result}")
print(f"等价于: NO(T, NO(J, NO(W, L))) = {NO(T, NO(J, NO(W, L)))}")
print(f"相等: {result == NO(T, NO(J, NO(W, L)))}")
```

## 2. 特殊情况

### 空参数列表

```python
result = normal_product()
print(f"normal_product() = {result}")
print(f"等于 One: {result == One}")
```

### 单个算符

```python
result = normal_product(T)
print(f"normal_product(T) = {result}")
print(f"等于 T: {result == T}")
```

### 包含零算符

```python
result1 = normal_product(T, Zero, J)
result2 = normal_product(Zero, T, J)

print(f"normal_product(T, Zero, J) = {result1}")
print(f"normal_product(Zero, T, J) = {result2}")
print(f"都等于 Zero: {result1 == Zero and result2 == Zero}")
```

### 包含单位算符

```python
result = normal_product(T, One, J)
expected = NO(T, J)

print(f"normal_product(T, One, J) = {result}")
print(f"等价于 NO(T, J) = {expected}")
print(f"相等: {result == expected}")
```

## 3. 包含导数算符

```python
result = normal_product(d(T), J, T)
expected = NO(d(T), NO(J, T))

print(f"normal_product(∂T, J, T) = {result}")
print(f"等价于 NO(∂T, NO(J, T)) = {expected}")
print(f"相等: {result == expected}")
```

## 4. 包含标量系数

```python
c = sp.Symbol("c")
k = sp.Symbol("k")

result = normal_product(c * T, k * J)
print(f"normal_product(c·T, k·J) = {result}")

# 标量系数会自动提取出来
expected = c * k * NO(T, J)
print(f"等价于 c·k·NO(T, J) = {expected}")
```

## 5. 实际应用示例：Virasoro 代数

```python
# 定义 Virasoro 生成元
T = BasicOperator("T", 2)

# 构建复合算符 (T(T(TT)))
T4 = normal_product(T, T, T, T)
print(f"T⁴ (嵌套正规序) = {T4}")

# 构建 (∂T(T(TT)))
dT_T3 = normal_product(d(T), T, T, T)
print(f"∂T·T³ (嵌套正规序) = {dT_T3}")
```

## 6. 与手动构建的对比

```python
# 使用 normal_product
auto = normal_product(T, J, W, L)

# 手动构建
manual = NO(T, NO(J, NO(W, L)))

print(f"自动构建: {auto}")
print(f"手动构建: {manual}")
print(f"\n相等: {auto == manual}")
print(f"\n使用 normal_product 更简洁！")
```

## 7. 多个算符（6个以上）

```python
# 创建 6 个算符
ops = [BasicOperator(f"O{i}", 1) for i in range(6)]

result = normal_product(*ops)
print(f"normal_product(O0, O1, O2, O3, O4, O5) = {result}")

# 手动构建会非常繁琐
manual = NO(ops[0], NO(ops[1], NO(ops[2], NO(ops[3], NO(ops[4], ops[5])))))
print(f"\n手动构建: {manual}")
print(f"\n相等: {result == manual}")
```

## 总结

`normal_product()` 函数的优势：

1. **简洁性**：`normal_product(A, B, C, D)` 比 `NO(A, NO(B, NO(C, D)))` 更简洁
2. **可读性**：直接表达算符乘积的意图
3. **可扩展性**：支持任意数量的算符，使用 `*ops` 解包
4. **自动处理**：自动处理零算符、单位算符和标量系数

### 使用场景

- 构建复杂的复合算符
- W-代数中的高阶生成元
- Sugawara 构造
- 任何需要多个算符正规序乘积的场合
