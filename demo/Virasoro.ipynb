```python
from pyope import *
from sympy import Symbol
```

```python
# 定义基本算符
T = BasicOperator("T", conformal_weight=2)  # Virasoro 应力张量，conformal weight = 2

c = Symbol('c')  # 中心荷

# 注册 OPE，用方括号 `[, ]`
OPE[T, T] = OPE.make(
    [c/2 * One, 0, 2 * T, d(T)]
)

basis = LocalOperatorBasis([T], max_weight=8)
basis.list(8)
```
