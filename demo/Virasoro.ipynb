```python
# 导入必要的库
import itertools
import sympy
from fractions import Fraction
from pyope import BasisOperator, OPE, simplify, NO, d, Bosonic, Fermionic, One
from sympy import Rational
import sys
sys.path.insert(0, '../src')
```

```python
# 定义基本算符
T = BasisOperator("T", h=2)  # Virasoro 应力张量，conformal weight = 2

c = sympy.Symbol('c')  # 中心荷

# 注册 OPE，用方括号 `[, ]`
OPE[T, T] = OPE.make(
    [c/2 * One, 0, 2 * T, d(T)]
)
```

```python
simplify(OPE(T, NO(T, T)).pole(1))
```
